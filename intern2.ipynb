```python
## Phase 2
## KPI Analysis
total_sales=df['Sales'].sum()

total_profit=df['Profit'].sum()

total_orders=df['Order ID'].nunique()

total_customers=df['Customer ID'].nunique()

avg_sales=df['Sales'].mean()
```

```python
kpi_table=pd.DataFrame(
    {
        'KPI': [
            'Total Sales',
            'Total Profit',
            'Total Orders',
            'Total Customers',
            'Average Sales'
        ],
        'Value':[
            total_sales,
            total_profit,
            total_orders,
            total_customers,
            avg_sales
        ]
    }
)


kpi_table['Value']=kpi_table['Value'].round(2)
kpi_table
```

```python
## Phase 3
## Sales Analysis
year_sales=df.groupby('Year')['Sales'].sum()
```

```python
year_sales.plot(kind='bar')
```

--Insight
-Year 2014 generated the highest sales and majorly contributed for revenue upliftment.
-Year 2011 and 2013 has slighter, yet distinguishable upliftment in sales.
-Year 2012 and 2011 has a slight decline, year 2012 generated the least sales.

```python
monthly_sales=df.groupby('Month')['Sales'].sum()
```

```python
monthly_sales = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

monthly_sales.index = [
    'Jan','Feb','Mar','Apr','May','Jun',
    'Jul','Aug','Sep','Oct','Nov','Dec'
]

monthly_sales.plot(kind='line', marker='o')
```

--Insight
-November generated the highest sales indicating strong seasonal demand while February generated the least sales.

# Phase 9: Business Insights & Recommendations

## Key Findings

### 1. Regional Performance
- West Region generated the highest sales.
- East Region showed strong profitability.

### 2. Product Performance
- Technology category generated the highest sales.
- Some sub-categories produced negative profits.

### 3. Customer Analysis
- Consumer segment contributed the highest revenue.

### 4. Discount Impact
- Higher discounts were associated with lower profits.

## Recommendations

1. Reduce excessive discounts on low-profit products.
2. Focus marketing on high-profit categories.
3. Improve inventory planning during peak sales months.
4. Review pricing strategy for loss-making products.
